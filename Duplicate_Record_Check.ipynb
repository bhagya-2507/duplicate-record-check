```python

import pandas as pd
import glob

# Excel files find karo
files = glob.glob("*.xlsx") + glob.glob("*.xls")

print("Files found:", files)
```

```python

df = pd.read_excel(files[0])

print("Dataset loaded successfully!")
print(df.head())
print("Rows and Columns:", df.shape)
```

```python

# Check duplicate rows

duplicate_count = df.duplicated().sum()

print("Total Duplicate Records:", duplicate_count)
```

```python

# Display duplicate records

duplicates = df[df.duplicated(keep=False)]

print(duplicates)
```

```python

# Create duplicate report

duplicate_count = df.duplicated().sum()

report = pd.DataFrame({
    "Check": ["Complete Duplicate Records"],
    "Total Duplicates": [duplicate_count],
    "Status": ["No Duplicates Found" if duplicate_count == 0 else "Duplicates Found"]
})

print(report)
```

```python

report.to_excel("Duplicate_Report.xlsx", index=False)

print("Duplicate report saved successfully!")
```

```python

# Remove complete duplicate rows
df_cleaned = df.drop_duplicates()

# Save cleaned dataset
df_cleaned.to_excel("Cleaned_Superstore.xlsx", index=False)

print("Cleaned dataset saved successfully!")
print("Original rows:", len(df))
print("Cleaned rows:", len(df_cleaned))
```
